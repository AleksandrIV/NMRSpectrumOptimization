#### Create model to optimize spin system

```python
model = model_single(spin_system)
```

### Prepare experimental spectrum

```python
exp_file = f'data/csv/{metabolite}.csv'
target = exp_spectrum_from_csv(exp_file, model.models[0].freq.numpy()) #We must know the frequency of the experimental spectrum spectrometer!!!
x = target[:,0].numpy()/ model.models[0].freq.numpy()
y = target[:,1].detach().numpy()
```

#### Spectra before optimization

```python
f=go.Figure(layout={'xaxis':{'title':'ppm'}})
f.add_trace(go.Scatter(x=x,y=target[:,1].detach().numpy(),name='target'))
f.add_trace(go.Scatter(x=x,y=model(target[:,0])[:,1].detach().numpy(),name='initial'))
```
